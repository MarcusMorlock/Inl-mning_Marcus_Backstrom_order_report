Individuell inlämningsuppgift (IG/G) – Fördjupning i Pythonprogrammering

Case: Refaktorera ett Pythonprogram för orderrapportering

Du har fått ta över ett mindre Pythonprogram som används av ett e-handelsföretag.
Programmet läser orderdata från en CSV-fil, bearbetar datan och skapar rapporter över bland annat försäljning och returer.
Programmet fungerar, men koden har flera problem:
stora delar av programmet ligger i samma fil,
olika delar av koden har otydliga ansvar,
vissa namn är svåra att förstå,
det finns duplicerad kod,
print() används för körinformation,
felhanteringen är begränsad,
det saknas automatiska tester,
delar av programmet är svåra att återanvända och testa.

----

Din uppgift är inte att bygga ett nytt analysverktyg.
Din uppgift är att ta den befintliga lösningen och göra den mer professionell, tydlig och möjlig att vidareutveckla.
Målet är att du ska visa att du kan arbeta med:
kodgranskning,
refaktorering,
projektstruktur och moduler,
logging,
objektorienterad programmering,
automatisk testning,
felhantering och validering,
dokumentation.

----

ad du ska göra
Arbeta stegvis. Det är ofta bättre att göra små förändringar och kontrollera att programmet fortfarande fungerar än att skriva om allt från början.

1. Förstå och granska originalkoden
Börja med att köra programmet och förstå vad det gör innan du förändrar det.
Skapa sedan filen:
code_review.md
Identifiera minst fem relevanta förbättringsområden i originalkoden.

För varje viktig punkt ska du beskriva:
Observation – vad ser du i koden?
Konsekvens – varför är det ett problem?
Förslag – hur skulle det kunna förbättras?

Exempel:
Observation: Samma funktion läser filen, bearbetar datan och sparar resultatet.
Konsekvens: Funktionen blir svår att testa eftersom varje test även behöver använda filsystemet.
Förslag: Separera filinläsning från transformationslogiken.

Fokusera på sådant som faktiskt påverkar kodens kvalitet, exempelvis:
ansvarsfördelning,
struktur,
testbarhet,
duplicering,
namngivning,
felhantering,
importer,
sidoeffekter,
logging,
hårdkodade värden eller sökvägar.

----

2. Strukturera om projektet
Dela upp koden i rimliga moduler med tydliga ansvar.
Du väljer själv exakt struktur.

Ett möjligt exempel är:

![Beskrivning](data/Image/cd8cb01de7ab5e3018c1f8e9a312aa56.png)

Du måste inte följa denna struktur exakt.

Det viktiga är att en annan programmerare kan förstå:
var programmet startar,
var data läses in,
var validering sker,
var data bearbetas,
var rapporterna skapas,
och var testerna finns.
Undvik att dela upp programmet i många små filer utan anledning. Fler filer betyder inte automatiskt bättre kod.

----

3. Använd moduler och importer på ett tydligt sätt
Din lösning ska använda fungerande Pythonmoduler och importer.
Undvik bland annat:
sys.path-lösningar för att få importer att fungera,
wildcard-importer som from module import *,
kod som startar hela programmet bara för att en modul importeras.
Programmet ska ha en tydlig startpunkt, exempelvis en main()-funktion eller motsvarande.

----

4. Använd logging
Ersätt print() med Pythons logging där utskrifterna beskriver programmets körning.
I moduler används normalt:

import logging

logger = logging.getLogger(__name__)

Logging ska konfigureras centralt i programmet.
Logga sådant som faktiskt hjälper någon att förstå vad som händer, exempelvis:
att data läses in,
hur många rader som lästs,
att validering genomförts,
att rapporter skapats,
var resultat sparats,
relevanta varningar eller fel.
Logging ska inte användas på varje rad och ersätter inte tydlig kod.

----

5. Använd OOP eller dataclass där det förbättrar designen
Din lösning ska innehålla minst en relevant klass eller dataclass.
Det är inte antalet klasser som bedöms.
Du ska använda objektorientering när det ger ett tydligare ansvar eller gör programmet enklare att använda och vidareutveckla.
Exempel på rimliga användningsområden kan vara:
programkonfiguration,
rapportinställningar,
valideringsresultat,
ett tydligt avgränsat objekt med både data och beteende.

Exempel - inte en mall du måste följa:
from dataclasses import dataclass
from pathlib import Path

@dataclass(frozen=True)
class ReportConfig:
    input_path: Path
    output_dir: Path

Undvik att skapa en stor klass som innehåller hela programmet bara för att uppfylla OOP-kravet.

----

6. Lägg till automatiska tester med pytest
Projektet ska innehålla automatiska tester med pytest.
Testerna ska i första hand testa din egen logik, inte att externa bibliotek som Pandas fungerar.
Du kan exempelvis testa:
beräkning av order_value,
beräkning efter rabatt,
hantering av returer,
sammanfattningar per kategori,
sammanfattningar per region,
saknade obligatoriska kolumner,
tom data,
felaktiga eller orimliga värden.
Du ska testa både:
normalfall – data som programmet förväntas kunna hantera,
och minst ett relevant:
edge case/felscenario – exempelvis saknad kolumn, tom data eller ogiltigt värde.
Testa inte bara att programmet ”går att köra”. Testerna ska kontrollera konkreta resultat eller förväntade fel.

----

7. Förbättra felhantering och validering
Fundera på vilka problem som rimligen kan uppstå.
Exempel:
- datafilen saknas,
- obligatoriska kolumner saknas,
- datan är tom,
- numeriska kolumner innehåller felaktiga värden,
- orimliga värden förekommer.
Du behöver inte bygga ett avancerat valideringssystem.
Målet är att fel ska upptäckas på ett tydligt sätt och ge information som hjälper användaren eller utvecklaren att förstå vad som hänt.
Undvik breda try/except som bara döljer felet.

----

8. Dokumentera projektet
Skapa en:
README.md
Den ska göra det möjligt för någon annan att förstå hur projektet används.
README ska kort beskriva:
- vad projektet gör,
- hur projektet förbereds eller installeras,
- vilka beroenden som behövs,
- hur programmet körs,
- hur testerna körs,
- projektets huvudsakliga struktur.
README behöver inte vara lång.

----

Reflektion
Avsluta med en kort reflektion i README.md eller i en separat reflection.md.
Besvara kort:
- 1: Vilka var de viktigaste problemen i originalkoden?
- 2: Vilka förändringar tycker du förbättrade programmet mest?
- 3: Varför valde du den projektstruktur du använde?
- 4: Var använde du OOP/dataclass och varför passade det där?
- 5: Vilka viktiga beteenden skyddar dina automatiska tester, och vilken nytta ger testerna om programmet förändras i  framtiden?
- 6: Vad var svårast?
- 7: Vad hade du velat förbättra ytterligare om du haft mer tid?
